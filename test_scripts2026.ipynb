```python
def func():
    word = input()  # Запрос слова у пользователя
    letter = input()  # Запрос буквы у пользователя
    word = word.lower()
    letter = letter.lower()

    word_string = list(word)
    count = 0
    for jet in word_string:
        if jet == letter:
            count = count + 1
            
    return count

# Выводим результат подсчета
print(func())
```

```python
# input: DDSSSAW
# output: D2S3AW

word = "DDSSSAW"
seen = ""
for c in word:
    if c not in seen:
        if word.count(c) == 1:
            seen = seen + c
        else:
            seen = seen + c + str(word.count(c))

print(seen)
#for cnt in input_string:
    
```

```python
def divide(x, y):
    if y == 0:
        raise ValueError("Деление на ноль не допускается.")
    return x / y

try:
    result = divide(10, 5)
    print(result)
except ValueError as e:
    print(e)
```

```python
def calculate_average(numbers):
    sum = 0
    for num in numbers:
        sum += num
    return sum/len(numbers)

numbers = [1,2,3,4,5]
result = calculate_average(numbers)
print(result)
```
