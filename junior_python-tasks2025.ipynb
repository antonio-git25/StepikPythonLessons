```python
#python tasks from geekbrain: task#8
def is_palindrome(string):
    return string == ''.join(reversed(string))

def is_polindrom(in_string):
    right = str(in_string)
    left = list(in_string)
    left.reverse()

    left2 = ''
    for i in left:
        left2 = left2 + i
    
    if right == left2:
        print(in_string, ' is polyndrom ')
    else:
        print(in_string, ' is not polyndrom ')

is_polindrom('redft')
is_polindrom('pop')
```

```python
#python tasks from geekbrain: task#9

def input_sec(sec):
    day = int(sec/(3600*24))
    hur = int((sec - (day * 3600 * 24))/3600)
    mint = int((sec - (day*3600*24+hur*3600))/60)
    secs = sec - (day*3600*24+hur*3600+mint*60)
    print('Days:', day)
    print('Hours:', hur)
    print('Minutes:', mint)
    print('Seconds:', secs)
    

input_sec(200000)

    
```
